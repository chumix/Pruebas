```python
# Paso 1: Instalación de Whisper y sus dependencias
print("Instalando Whisper...")
!pip install -q openai-whisper  # Añadido -q para minimizar el output
print("Whisper instalado.")

# Importar librerías
import os
import glob
from google.colab import drive

# Paso 2: Configuración inicial
# Montar Google Drive
print("Montando Google Drive...")
drive.mount('/content/drive')
print("Google Drive montado.")

# Definir la carpeta en Google Drive que contiene los archivos de audio
input_folder = '/content/drive/MyDrive/03.AUDIOS/Temp_sync/aleman'
output_folder = '/content/drive/MyDrive/03.AUDIOS/Temp_sync/aleman'  # Guardar en una subcarpeta de salida

# Crear la carpeta de salida si no existe
os.makedirs(output_folder, exist_ok=True)

# Verificar que la carpeta de entrada existe
if not os.path.exists(input_folder):
    raise FileNotFoundError(f"La carpeta {input_folder} no existe")
print(f"Carpeta de entrada verificada: {input_folder}")

# Obtener la lista de archivos .m4a en la carpeta
print("Obteniendo lista de archivos .m4a...")
audio_files = glob.glob(os.path.join(input_folder, '*.m4a'))
total_files = len(audio_files)
print(f"Se encontraron {total_files} archivos de audio.")

# Paso 3: Transcribir cada archivo usando el comando whisper
for idx, audio_file in enumerate(audio_files):
    print(f"\nProcesando archivo {idx + 1} de {total_files}: {audio_file}")

    # Construir el comando para ejecutar Whisper
    command = f'whisper "{audio_file}" --task transcribe --model tiny --verbose True --output_dir "{output_folder}"'

    # Ejecutar el comando
    print(f"Ejecutando el comando: {command}")
    os.system(command)
    print(f"Transcripción completada para: {audio_file}")

    # Mostrar el porcentaje de progreso
    progress = (idx + 1) / total_files * 100
    print(f"Progreso: {progress:.2f}%")

print("\nProceso completado.")
```
